# Projection

This notebook accompanies Chapter 2, section on the inner product and projection.

A uniform load $f = 2$ deflects the string into the parabola $u(x) = x(1-x)$, and one blend of the three modes is closest to it in the $L^2$ distance.
Closeness by perpendicularity needs angles, and the inner product $\langle f, g \rangle = \int_0^1 f g \, dx$ supplies them.
The first lab rotates one unit function against another and shows the inner product as net signed overlap.
Green bars mark where the two functions agree in sign, red where they oppose, and at $90$ degrees the two areas cancel exactly.

The cell below reproduces the section's measured values, the projected coefficients, the orthogonality of the error, the Pythagorean split, and the cost of a mistuned coefficient.


```python
import numpy as np

x = np.linspace(0.0, 1.0, 8001)
parab = x * (1 - x)
modes = [np.sin(k * np.pi * x) for k in (1, 2, 3)]

def inner(f, g):
    return np.trapezoid(f * g, x)

c = [2 * inner(parab, m) for m in modes]
best = sum(ck * m for ck, m in zip(c, modes))
e = parab - best
print(f"c1 = {c[0]:.4f}  (8/pi^3   = {8/np.pi**3:.4f})")
print(f"c2 = {c[1]:.1e}")
print(f"c3 = {c[2]:.5f} (8/27pi^3 = {8/(27*np.pi**3):.5f})")
print("error orthogonal to each mode:", ", ".join(f"{abs(inner(e, m)):.1e}" for m in modes))
print(f"Pythagoras: ||u||^2 = {inner(parab, parab):.6f} = "
      f"{inner(best, best):.6f} + {inner(e, e):.2e}")
print(f"relative L2 error of the best blend: {np.sqrt(inner(e, e)/inner(parab, parab)):.2%}")
mist = 1.2 * c[0] * modes[0] + c[2] * modes[2]
em = parab - mist
print(f"c1 mistuned +20%: error norm ratio = "
      f"{np.sqrt(inner(em, em)/inner(e, e)):.1f}x")
```

## A Cauchy sequence leaving the continuous functions

A finite-dimensional family contains its own closest point.
For families with infinitely many shapes, a sequence can improve forever, and the lab below runs one to the edge of a space.
Each stage halves the width of a continuous ramp; the members approach one another and the $L^2$ limit is the step, which is not continuous.
A space in which every such sequence keeps its limit inside is complete, and a complete inner-product space is a Hilbert space.

## A difference at one point weighs nothing

The second lab squeezes a spike of fixed height toward a single point.
The $L^2$ norm falls toward zero while the peak stays at one, so the limit is the zero function in $L^2$, and a difference at exactly one point has distance zero.
An element of $L^2$ has no value at a point, which is why sensor readings need the finite-energy space $H^1$.

## What comes next

$L^2$ and $H^1$ are complete inner-product spaces, Hilbert spaces, and the projection theorem's closest point exists in them.
